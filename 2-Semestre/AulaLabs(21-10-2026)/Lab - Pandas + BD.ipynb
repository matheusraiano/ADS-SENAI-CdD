# Importando as bibliotecas do projeto

```python
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass
```

# Credencias do banco de dados

```python
usuario = getpass("Usuário: ")
senha = getpass("Senha: ")
endpoint = getpass("EndPoint: ")
```

```python
url = f"mysql+pymysql://{usuario}:{senha}@{endpoint}:3306/mysql"
print(url)
```

# Primeira conexão com o banco de dados

```python
engine = create_engine(url)
with engine.begin() as conexao:
  print("Conexão realizada com sucesso!")
```

# Criando uma tabela no banco de dados

```python
with engine.begin() as conexao:
  conexao.exec_driver_sql(
      "CREATE DATABASE teste"
      )
  print("Banco de dados criado com sucesso!")
```

# Enviando o dataframe para o bando de dados (Nuvem)

```python
tabela = {
    "sensor":['S01', 'S02', 'S03'],
    "setor":['Usinagem', "Montagem", "Pintura"],
    "leituras":[72.5,81.3,32.6]
} # Criando uma tabela manualmente
dados = pd.DataFrame(tabela)
dados.head()
```

```python
url_database_teste = f"mysql+pymysql://{usuario}:{senha}@{endpoint}:3306/teste"
engine = create_engine(url_database_teste)
dados.to_sql("sensores", engine, if_exists="replace", index=False)
```

## Verificando se os dados foram salvos na nuvem

```python
teste = pd.read_sql("SELECT * FROM sensores", engine)
teste.head()
```
