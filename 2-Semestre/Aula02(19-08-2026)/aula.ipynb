```python
import numpy as np

exemplo = pd.DataFrame({
    'temperatura': [70, 72, np.nan, 74, 76],
    'tempo_ciclo': [10, 11, np.nan, 12, 50],
    'turno': ['manha', 'manha', np.nan, 'tarde', 'manha']
})
exemplo
```

```python
exemplo.fillna(
    {
       'temperatura' : exemplo['temperatura'].mean(),
      'tempo_ciclo' : exemplo['tempo_ciclo'].median(),
      'turno' :  exemplo['turno'].mode()[0]
    }, inplace=True
)
exemplo
```

#Média / Mediana

```python
exemplo['temperatura'].mean()
```

```python
media = exemplo['temperatura'].mean()
exemplo.fillna(
    {
        'temperatura' : media
    }, inplace=True
)
```

```python
mediana = exemplo['tempo_ciclo'].median()
exemplo.fillna(
    {
        'tempo_ciclo' : mediana
    }, inplace=True
)
```

```python
exemplo.isna().sum()
```

#Moda

```python
exemplo['turno'].mode()[0]
```

```python
moda = exemplo['turno'].mode()[0]
exemplo.fillna(
    {
        'turno' : moda
    }, inplace=True
)
```

```python
exemplo.isna().sum()
```

#Analise de frquencia e agrupamentos

```python
dados['Churn'].value_counts()
```

```python
dados.groupby('Contract')['TotalCharges'].mean()
```

```python
tab = dados.groupby('Contract')['TotalCharges'].sum()
tab.sort_values(ascending=False)
```

```python
tab_2 = dados.groupby('Contract', as_index=False)['MonthlyCharges'].mean()
tab_2.sort_values(by='MonthlyCharges', ascending=False)
```

```python
dados.groupby('Churn').size()
```

```python
dados.groupby('Contract').agg(
    # variavel = ('Nome_Coluna', 'Metrica')
    qtde = ('customerID', 'count'),
    media_mensal = ('MonthlyCharges', 'mean'),
    total = ('TotalCharges', 'sum'),
    menor_total = ('TotalCharges', 'min'),
    maior_total = ('TotalCharges', 'max')
)
```

```python
dados.groupby(['Contract', 'Churn']).agg(
    # variavel = ('Nome_Coluna', 'Metrica')
    qtde = ('customerID', 'count'),
    media_mensal = ('MonthlyCharges', 'mean')
)
```

#Graficos

```python
import matplotlib.pyplot as plt
```

```python
dados['Churn'].value_counts().plot.bar()
plt.title('Proporção de Desistências')
plt.xlabel('Rotatividade de Clientes')
plt.ylabel('Frequência de Ocorrências')
plt.show()
```

```python
tabela = dados.groupby('Contract')['TotalCharges'].sum()
tabela.plot.bar()
plt.title('Total de Gastos por Contrato')
plt.xlabel('Tipo de Contrato')
plt.ylabel('Total Gasto (x100000 $)')
plt.show()
```

```python
tabela = dados.groupby('Churn')['MonthlyCharges'].mean()
tabela.plot.bar()
plt.title('Média de Gastos por Rotatividade')
plt.xlabel('Rotatividade')
plt.ylabel('Média de Gastos')
plt.show()
```
