# Problema 02: Valores diferentes representando a mesma informação (`situacao`)

```python
dados['situacao'].value_counts()
```

```python
mapa_situacao = {
    # Operando
    'Operando' : 'Operando', # isso é para não deixar NaN, pois se o map não saber o que fazer com um dado ele deixa NaN
    'OPERANDO' : 'Operando',
    'operando' : 'Operando',
    'Em operação' : 'Operando',
    'Ligado' : 'Operando',
    # Parado
    'Parado' : 'Parado', # isso é para não deixar NaN, pois se o map não saber o que fazer com um dado ele deixa NaN
    'parado' : 'Parado',
    'PARADO' : 'Parado',
    'Fora de operação' : 'Parado',
    'Desligado' : 'Parado',
    # Manutenção
    'Manutenção' : 'Manutenção', # isso é para não deixar NaN, pois se o map não saber o que fazer com um dado ele deixa NaN
    'Em manutenção' : 'Manutenção',
    'manutencao' : 'Manutenção',
    'MANUTENÇÃO' : 'Manutenção',
    'Manut.' : 'Manutenção'
}
```

```python
dados['situacao'].map(mapa_situacao).isna().sum()
```

```python
# ver qual é o erro que pode ter dado(erro = np.int64(0) difetente de 0)
mask = dados['situacao'].map(mapa_situacao).isna()
dados[mask]['situacao'].unique()
```

```python
dados['situacao'] = dados['situacao'].map(mapa_situacao)
```

```python
dados['situacao'].unique()
```
