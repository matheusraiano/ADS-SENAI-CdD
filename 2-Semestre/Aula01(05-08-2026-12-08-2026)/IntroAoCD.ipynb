##3.1. Filtrando Linhas e Colunas

### 3.1.1. Mascaras --> Filtro de Linhas

```python
mask = dados['alcohol'] >= 9.8
dados[mask]
```

```python
mask = dados['quality'] > 5
dados[mask]
```

###3.1.2. Operadores Lógicos


1.   AND --> &
2.   OR --> |



```python
mask = (dados['quality'] == 5) & (dados['alcohol'] >= 9.8)
dados[mask]
```

###3.2.3. Usando Dois Filtros

```python
mask = (dados['quality'] == 5) & (dados['alcohol'] >= 9.8)
dados[mask][['quality', 'alcohol']]
```

### 3.2.5. Criando Novas Colunas

```python
dados['exemplo'] = 5
dados.head()
```

```python
dados['exemplo'].unique()
```

```python
dados['quality'].unique()
```

```python
dados['exemplo_1'] = dados['pH'] + dados['density']
dados[['pH','density','exemplo_1']]
```

```python
# Nova Coluna Categórica a partir de uma quantitativa
dados['faixa_alcool'] = 'medio'
# mascaras para filtrar e classificar:
mask_baixo = dados['alcohol'] < 9
mask_alto = dados['alcohol'] >= 11
dados.loc[mask_baixo, 'faixa_alcool'] = 'baixo'
dados.loc[mask_alto, 'faixa_alcool'] = 'alto'
dados['faixa_alcool'].unique()
```

### 3.2.6 Excluindo Colunas da Tabela

```python
dados = dados.drop(columns=['exemplo', 'exemplo_1'])
```

```python
dados.drop(columns=['exemplo', 'exemplo_1'], inplace=True)
```
