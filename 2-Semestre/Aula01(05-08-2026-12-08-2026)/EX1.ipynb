# Parte 3 — Filtros

## Filtro 1 — Temperatura elevada

```python
mask = dados['temperatura_c'] >= 80
dados[mask]
```

## Filtro 2 — Máquinas em alerta

```python
mask = dados['status_maquina'] == "Alerta"
dados[mask]
```

## Filtro 3 — Setor específico

```python
mask = dados['setor'] == "Usinagem"
dados[mask][['setor']]
```

## Filtro 4 — Temperatura crítica na Usinagem

```python
mask = (dados['setor'] == "Usinagem") & (dados['temperatura_c'] >= 85)
dados[mask]
```

## Filtro 5 — Problemas durante o turno da noite

```python
mask = (dados['turno'] == "Noite") & (dados['vibracao_mm_s'] >= 5.5)
dados[mask]
```

## Filtro 6 — Máquinas paradas

```python
mask = (dados['status_maquina'] == "Parada") & (dados['tempo_parada_min'] >= 30)
dados[mask]
```

## Filtro 7 — Alta produção

```python
mask = dados['producao_unidades'] >= 300
dados[mask][['maquina', 'setor', 'turno', 'producao_unidades']]
```

## Filtro 8 — Produção com poucos defeitos

```python
mask = (dados['producao_unidades'] >= 280) & (dados['defeitos'] == 0)
dados[mask]
```

## Filtro 9 — Condição elétrica de atenção

```python
mask = (dados['corrente_a'] >= 22) & (dados['temperatura_c'] >= 82)
dados[mask][['maquina', 'temperatura_c', 'corrente_a', 'status_maquina']]
```

## Filtro 10 — Condição por dois setores

```python
mask = (dados['setor'] == "Montagem") | (dados['setor'] >= "Pintura")
dados[mask]
```

# Parte 5 — Criando uma classificação

```python
dados['faixa_temperatura'] = "Normal"
mask_atencao = dados['temperatura_c'] >= 75
mask_critica = dados['temperatura_c'] >= 90
dados.loc[mask_atencao, 'faixa_temperatura'] = 'Atencao'
dados.loc[mask_critica, 'faixa_temperatura'] = 'Critica'
dados['faixa_temperatura'].unique()
```

# Parte 6 — Resumindo os dados
